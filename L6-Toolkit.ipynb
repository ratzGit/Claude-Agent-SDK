```python
# Import list formatted for clarity
from claude_agent_sdk import (
    query, ClaudeAgentOptions, 
    ResultMessage, AssistantMessage, 
    ToolUseBlock, ThinkingBlock, TextBlock
)
# Importing classes to format the output in the desired way
from IPython.display import display, Markdown

# Declare and initialize variables to keep the logic clean
prompt = "In the current directory search for README.md and show it in markdown format"
options = ClaudeAgentOptions(
    model="claude-haiku-4-5",
    thinking={"type":"enabled", "budget_tokens":1000, "display":"summarized"},
)
```

```python
async def main():
    # Use the for loop to iterate the output of the query()
    async for msg in query(prompt=prompt,options=options):
        # Check if the message yielded is a ResultMessage instance
        if isinstance(msg, ResultMessage):
                display(Markdown(msg.result))
        # Check if the message yielded is an AssistantMessage instance
        elif isinstance(msg, AssistantMessage):
                # If yes, iterate the contents using a for loop
                for block in msg.content:
                      # Check the type of the block
                      if isinstance(block, ToolUseBlock):
                            # If yes, print the name of the tool used
                            print(f"Tool used is: {block.name}")
                      elif isinstance(block, ThinkingBlock):
                            # If yes, print what the agent is thinking or at least confirm if there is a signature
                            print(f"Agent is thinking: {block.thinking}")
                      elif isinstance(block, TextBlock):
                            # If yes, just print the content text generated
                            print(f"Agent generated text: {block.text}")
```

```python
await main()
```
