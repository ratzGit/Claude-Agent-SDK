```python
# Program to enforce a required model and print the model usage statistics.
from claude_agent_sdk import query, ClaudeAgentOptions, ResultMessage

options = ClaudeAgentOptions(
        model="claude-haiku-4-5",
        # Set the environment variable to enforce the use of the specified model
        env={"ANTHROPIC_DEFAULT_HAIKU_MODEL": "claude-haiku-4-5"}
        )
```

```python
async def main():    
    async for message in query(
        prompt="Say hello world in a quirky way.",  
        options=options 
    ):
        if isinstance(message, ResultMessage):
            for key, value in message.model_usage.items():
                print(f"{key}: {value}")
```

```python
await main()
```
