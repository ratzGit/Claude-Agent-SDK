```python
from claude_agent_sdk import (
    query, ClaudeAgentOptions,
    SystemMessage, UserMessage, AssistantMessage, ResultMessage,
    TextBlock, ThinkingBlock, ToolUseBlock, ToolResultBlock,
)

# Using a prompt that is simple but needs some thinking and tool calls
prompt = """Write a three line comment about the contents 
of this folder in a new file called Observations.txt"""
options=ClaudeAgentOptions(
     model = "claude-haiku-4-5",
     thinking={"type":"enabled","budget_tokens":1000 ,"display":"summarized"}
)


async def main():
        async for msg in query(
            prompt=prompt,
            options=options,
        ):

            # Using Match-Case to avoid a maze of if else conditions
            match msg:
                case SystemMessage():
                    # Simply print out the sub-type
                    print(f"SystemMessage sub-type: {msg.subtype}")
                case AssistantMessage():
                    print("Entered AssistantMessage \n")
                    # Iterate and check if its a response, thinking or tool call
                    for block in msg.content:
                        if isinstance(block, TextBlock):
                            print(f"Claude's response: {block.text}")
                        elif isinstance(block, ThinkingBlock):
                            print(f"Claude's response: {block.thinking}")
                        elif isinstance(block, ToolUseBlock):
                            print(f"Claude requested to use {block.name} tool")
                case UserMessage():
                    print("Entered UserMessage \n")
                    # UserMessage can have direct string response or a block as response
                    if msg.content == str:
                        print(msg.content)
                    else:
                        for block in msg.content:
                            if isinstance(block, TextBlock):
                                print(block.text)
                            elif isinstance(block, ToolResultBlock):
                                print(f"ToolUseResult: {block.content}")
                case ResultMessage():
                    print(f"\n The final result is: {msg.result} And the total cost was {msg.total_cost_usd}")
                    
                           

await main()
```
