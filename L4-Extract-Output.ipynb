```python
# Import the query function from the claude_agent_sdk module
from claude_agent_sdk import query, SystemMessage
```

```python
# Define an asynchronous main function to handle the query
async def main():
    async for message in query(
        prompt="Say hello world in a quirky way.",       
    ):
        if isinstance(message, SystemMessage):
            print(message.data.get("cwd"))
```

```python
# Call the async function directly using top-level await
await main()
```
